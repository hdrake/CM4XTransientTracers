```python
for modelexp in grids.keys():
    inv_path = f"../data/interim/transient_tracer_inventory_{modelexp}.nc"

    ds = grids[f"{modelexp}"]._ds
    Δt = ds.time.dt.days_in_month * sec_per_day

    ds["volcello"] = ds.thkcello.fillna(0.)*ds.areacello.fillna(0.)

    print(f"Computing globally-integrated inventory for {modelexp}: ")
    inv = xr.Dataset()

    for tr in tracers:
        print(tr, end=", ")
        
        inv[f'{tr}_volumeint'] = (ds[tr]*ds['volcello']).sum(['xh', 'yh', 'z_l']).compute()
        inv[f'{tr}_peak_year'] = np.int64(inv[f'{tr}_volumeint'].idxmax('year').compute())

    print("")
    inv.to_netcdf(inv_path)
```
